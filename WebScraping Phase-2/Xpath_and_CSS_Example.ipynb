```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# Start Chrome browser
driver = webdriver.Chrome()

# Open website
driver.get("http://quotes.toscrape.com/")

# Wait until quotes load 
WebDriverWait(driver, 10).until(
    EC.presence_of_element_located((By.CSS_SELECTOR, "div.quote"))
)

# Get all quote blocks using CSS
quotes = driver.find_elements(By.CSS_SELECTOR, "div.quote")

for quote in quotes:
    
    # Get quote text using CSS Selector
    text = quote.find_element(By.CSS_SELECTOR, "span.text").text
    
    # Get author using CSS Selector
    author = quote.find_element(By.CSS_SELECTOR, "small.author").text
    
    # Get author profile link using XPath (relative XPath)
    author_link = quote.find_element(By.XPATH, ".//a[text()='(about)']").get_attribute("href")
    
    print("Quote:", text)
    print("Author:", author)
    print("Author Link:", author_link)
    print("-" * 50)

# Close browser
driver.quit()
```

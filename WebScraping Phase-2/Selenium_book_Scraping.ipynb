```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import csv
import time

# open chrome browser
driver = webdriver.Chrome()

# open the website
driver.get("https://books.toscrape.com/")

time.sleep(3)  # wait for page to load

books_data = []

# get all books
books = driver.find_elements(By.CLASS_NAME, "product_pod")

print("Collecting first 10 books...")

# take only first 10 books
for book in books[:10]:

    try:
        title = book.find_element(By.CSS_SELECTOR, "h3 a").get_attribute("title")
        price = book.find_element(By.CLASS_NAME, "price_color").text
        availability = book.find_element(By.CLASS_NAME, "instock").text.strip()

        rating_class = book.find_element(By.CLASS_NAME, "star-rating").get_attribute("class")
        rating = rating_class.split()[-1]

        url = book.find_element(By.CSS_SELECTOR, "h3 a").get_attribute("href")

        books_data.append([title, price, rating, availability, url])

    except:
        print("Skipping a book...")

# save to csv
with open("books.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["Title", "Price", "Rating", "Availability", "URL"])

    for row in books_data:
        writer.writerow(row)

print("Data saved to books.csv")

driver.quit()
```

```python
from sqlalchemy import create_engine, Column, Integer, String, Float
from sqlalchemy.orm import declarative_base, sessionmaker

# connect to PostgreSQL database
engine = create_engine("postgresql+psycopg2://postgres:seshu123@localhost:5432/scraper_db")

Base = declarative_base()

# create table structure
class Product(Base):
    __tablename__ = "products"

    id = Column(Integer, primary_key=True)
    name = Column(String)
    price = Column(Float)
    rating = Column(Float)

# create table if it doesn't exist
Base.metadata.create_all(engine)

# start session
Session = sessionmaker(bind=engine)
session = Session()

# sample scraped data
products_data = [
    {"name": "Laptop", "price": 55000, "rating": 4.5},
    {"name": "Mobile", "price": 20000, "rating": 4.2},
    {"name": "Headphones", "price": 1500, "rating": 4.0}
]

# insert data into table
for item in products_data:
    new_product = Product(
        name=item["name"],
        price=item["price"],
        rating=item["rating"]
    )
    session.add(new_product)

session.commit()

print("Data inserted successfully!")

# read data from database
all_products = session.query(Product).all()

for p in all_products:
    print(p.id, p.name, p.price, p.rating)

session.close()
```

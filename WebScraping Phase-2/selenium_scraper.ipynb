```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
import csv

# open chrome browser
driver = webdriver.Chrome()

# open the website
driver.get("http://quotes.toscrape.com/")

data = []

# we want to scrape first 3 pages
for i in range(3):

    print("Scraping Page:", i + 1)

    time.sleep(2)   # wait for page to load

    quotes = driver.find_elements(By.CLASS_NAME, "quote")

    # loop through each quote
    for q in quotes:
        quote_text = q.find_element(By.CLASS_NAME, "text").text
        quote_author = q.find_element(By.CLASS_NAME, "author").text

        data.append([quote_text, quote_author])

    # go to next page
    try:
        next_btn = driver.find_element(By.LINK_TEXT, "Next")
        next_btn.click()
    except:
        print("No more pages found")
        break


# save data into csv file
with open("quotes.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)

    writer.writerow(["Quote", "Author"])

    for row in data:
        writer.writerow(row)


print("Data saved successfully in quotes.csv")

driver.quit()
```
